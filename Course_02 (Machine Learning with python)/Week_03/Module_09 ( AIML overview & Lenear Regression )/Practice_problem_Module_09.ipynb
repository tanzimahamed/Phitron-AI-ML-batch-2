### Task 2

```python
# Simple equation ব্যবহার করে prediction করো
# y = mx + b


m = 300
b = 50000

predicted_price = (m * house_size) + b
print("Predicted Price:", predicted_price)
```

### Task 4

```python
# matplotlib ব্যবহার করে একটি straight line plot করো

import matplotlib.pyplot as plt
import numpy as np


x = np.array([1000, 1200, 1500, 1800, 2000])
m = 300
b = 50000
y = (m * x) + b


plt.plot(x, y, color='red', marker='o', label='Best Fit Line')
plt.title("House Price Prediction Line")
plt.xlabel("House Size (sq ft)")
plt.ylabel("Price (BDT)")
plt.grid(True)
plt.legend()
plt.show()
```

### Task 5

```python
# sklearn ব্যবহার করে একটি simple linear regression model তৈরি করো



from sklearn.linear_model import LinearRegression
import numpy as np

X = np.array([[1000], [1200], [1500], [1800], [2000]]) # House Sizes
y = np.array([350000, 410000, 500000, 590000, 650000]) # Actual Prices


model = LinearRegression()
model.fit(X, y)

new_house_size = [[1700]]
predicted_val = model.predict(new_house_size)

print(f"Predicted Price for 1700 sq ft house: {predicted_val[0]:.2f}")
print("Model Slope (m):", model.coef_[0])
print("Model Intercept (b):", model.intercept_)
```

1. Regression predicts ________

3. ML learns from ________
4. Cost Function measures ________
5. Deep Learning uses ________ Networks



# Ans:
Regression predicts continuous numerical values (e.g., house price, temperature).

Classification predicts categorical labels or discrete classes (e.g., Spam or Not Spam, Cat or Dog).

ML learns from historical data/patterns.

Cost Function measures the error or difference between predicted values and actual values.
