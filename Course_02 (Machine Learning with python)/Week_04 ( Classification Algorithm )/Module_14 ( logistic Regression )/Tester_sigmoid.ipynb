```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
def sigmoid ( z ):
  return 1/(1+np.exp(-z))
```

```python
z = np.arange(-10,10,0.1)
values = sigmoid(z)
plt.plot(z,values)
plt.xlabel('z')
plt.ylabel('sigmoid(z)')
plt.title('sigmoid function')
```
